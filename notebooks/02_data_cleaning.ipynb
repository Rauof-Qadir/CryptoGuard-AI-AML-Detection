```python
import pandas as pd
import numpy as np


features = pd.read_csv(
    "../data/raw/elliptic_txs_features.csv",
    header=None
)

classes = pd.read_csv(
    "../data/raw/elliptic_txs_classes.csv"
)

edges = pd.read_csv(
    "../data/raw/elliptic_txs_edgelist.csv"
)
```

```python
features.head()
```

```python
features.columns = (
    ['txId','time'] +
    [f'feature_{i}' for i in range(1,166)]
)
```

```python
features.shape
```

```python
features.isnull().sum().sum()
```

```python
classes.head()
```

```python
data = features.merge(
    classes,
    on='txId'
)
```

```python
data.head()
```

```python
data = data[
    data['class'] != 'unknown'
]
```

```python
data['class'] = data['class'].astype(int)
```

```python
data['class'].value_counts()
```

```python
from sklearn.preprocessing import StandardScaler


X = data.drop(
    ['txId','class'],
    axis=1
)


scaler = StandardScaler()


X_scaled = scaler.fit_transform(X)
```

```python
clean_data = pd.DataFrame(
    X_scaled,
    columns=X.columns
)


clean_data['class']=data['class'].values


clean_data.to_csv(
"../data/processed/elliptic_clean.csv",
index=False
)
```
